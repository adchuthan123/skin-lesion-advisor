# Notebook 2: ML Training — Risikoklassifikation aus Metadaten
**Skin Lesion Risk Advisor**

Dieses Notebook trainiert einen ML-Klassifikator auf Patientenmetadaten (Alter, Geschlecht, Körperstelle).
Der Output `risk_level` (low / medium / high) wird später mit dem CV-Output kombiniert.

**Iteration 1:** Basisfeatures (age, sex, localization)  
**Iteration 2:** + Feature Engineering (Altersgruppen, Körperstellen-Flags)  
**Modelle:** Random Forest vs. XGBoost

## 0. Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report, confusion_matrix,
    balanced_accuracy_score, f1_score
)
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBClassifier
import joblib
import warnings
warnings.filterwarnings('ignore')

FIGURES_DIR = Path('../reports/figures')
MODELS_DIR  = Path('../models')
MODELS_DIR.mkdir(exist_ok=True)

print('Setup OK')
```

## 1. Daten laden & Zielvariable

```python
df = pd.read_csv('../data/processed/metadata_clean.csv')
print(f'Geladen: {df.shape}')
df.head()
```

```python
# Zielvariable: risk_level basierend auf Diagnose
# high   = bösartige Läsionen (mel, bcc)
# medium = Vorstufen (akiec)
# low    = gutartige Läsionen (bkl, nv, vasc, df)
risk_map = {
    'mel':   'high',
    'bcc':   'high',
    'akiec': 'medium',
    'bkl':   'low',
    'nv':    'low',
    'vasc':  'low',
    'df':    'low'
}
df['risk_level'] = df['dx'].map(risk_map)

print('Risikoverteilung:')
print(df['risk_level'].value_counts())
```

---
## Iteration 1: Basisfeatures

**Features:** `age`, `sex` (encoded), `localization` (encoded)  
**Ziel:** Baseline-Performance mit den rohen Metadaten

```python
# Label-Encoding der kategorischen Features
le_sex  = LabelEncoder()
le_loc  = LabelEncoder()
le_risk = LabelEncoder()

df['sex_enc'] = le_sex.fit_transform(df['sex'])
df['loc_enc'] = le_loc.fit_transform(df['localization'])
df['risk_enc'] = le_risk.fit_transform(df['risk_level'])

print('Klassen risk_level:', le_risk.classes_)
print('Klassen sex:       ', le_sex.classes_)

# Encoders speichern (werden vom App verwendet)
joblib.dump(le_sex,  MODELS_DIR / 'le_sex.pkl')
joblib.dump(le_loc,  MODELS_DIR / 'le_loc.pkl')
joblib.dump(le_risk, MODELS_DIR / 'le_risk.pkl')
print('Encoders gespeichert.')
```

```python
FEATURES_V1 = ['age', 'sex_enc', 'loc_enc']

X_v1 = df[FEATURES_V1]
y    = df['risk_enc']

X_train, X_test, y_train, y_test = train_test_split(
    X_v1, y, test_size=0.2, random_state=42, stratify=y
)
print(f'Train: {X_train.shape} | Test: {X_test.shape}')
print(f'Features Iteration 1: {FEATURES_V1}')

# Split-Index für spätere Nutzung merken
test_idx = X_test.index
```

```python
# Random Forest — Iteration 1
rf_v1 = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
rf_v1.fit(X_train, y_train)
y_pred_rf_v1 = rf_v1.predict(X_test)

bal_rf_v1 = balanced_accuracy_score(y_test, y_pred_rf_v1)
f1_rf_v1  = f1_score(y_test, y_pred_rf_v1, average='macro')

print('=== Random Forest — Iteration 1 ===')
print(f'Balanced Accuracy: {bal_rf_v1:.3f}')
print(f'F1 (macro):        {f1_rf_v1:.3f}')
print()
print(classification_report(y_test, y_pred_rf_v1, target_names=le_risk.classes_))
```

```python
# XGBoost — Iteration 1
sw_v1 = compute_sample_weight('balanced', y_train)

xgb_v1 = XGBClassifier(
    n_estimators=200,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    random_state=42,
    eval_metric='mlogloss',
    verbosity=0
)
xgb_v1.fit(X_train, y_train, sample_weight=sw_v1)
y_pred_xgb_v1 = xgb_v1.predict(X_test)

bal_xgb_v1 = balanced_accuracy_score(y_test, y_pred_xgb_v1)
f1_xgb_v1  = f1_score(y_test, y_pred_xgb_v1, average='macro')

print('=== XGBoost — Iteration 1 ===')
print(f'Balanced Accuracy: {bal_xgb_v1:.3f}')
print(f'F1 (macro):        {f1_xgb_v1:.3f}')
print()
print(classification_report(y_test, y_pred_xgb_v1, target_names=le_risk.classes_))
```

---
## Iteration 2: Feature Engineering

**Neue Features (angelehnt an LN1 Apartment Prediction):**
- `age_group`: Altersgruppe (jung <30, mittel 30–60, alt >60)
- `is_face`: Binär — Läsion im Gesicht (erhöhtes Risiko für Basalzellkarzinom)
- `is_trunk`: Binär — Rücken / Bauch / Brust / Rumpf (häufig Melanom)
- `is_extremity`: Binär — Arme / Beine / Hände / Füsse
- `is_high_risk_location`: Binär — Rücken + Unterschenkel (häufigste Melanom-Stellen laut EDA)

**Ziel:** Domänenwissen aus der EDA als Features kodieren → bessere Klassen-Trennung

```python
# Altersgruppe mit numpy (vermeidet pd.cut NaN-Konvertierungsproblem)
df['age_group'] = np.where(df['age'] < 30, 0, np.where(df['age'] <= 60, 1, 2)).astype(int)

# Körperstellen-Flags (aus EDA Heatmap: Melanom häufig an Rücken & Unterschenkel)
df['is_face'] = df['localization'].isin(['face']).astype(int)

df['is_trunk'] = df['localization'].isin(
    ['back', 'abdomen', 'chest', 'trunk']
).astype(int)

df['is_extremity'] = df['localization'].isin(
    ['lower extremity', 'upper extremity', 'foot', 'hand']
).astype(int)

# Melanom-Risikostandorte laut EDA (Notebook 01 Heatmap)
df['is_high_risk_location'] = df['localization'].isin(
    ['back', 'lower extremity']
).astype(int)

FEATURES_V2 = [
    'age', 'sex_enc', 'loc_enc',          # Basisfeatures
    'age_group',                            # Neu: Altersgruppe
    'is_face', 'is_trunk',                  # Neu: Körperstellen-Flags
    'is_extremity', 'is_high_risk_location' # Neu: Risikostandort
]

print(f'Features Iteration 1: {len(FEATURES_V1)} Features — {FEATURES_V1}')
print(f'Features Iteration 2: {len(FEATURES_V2)} Features — {FEATURES_V2}')
print()
print('Verteilung is_high_risk_location:')
print(df.groupby('is_high_risk_location')['risk_level'].value_counts().unstack())
```

```python
X_v2 = df[FEATURES_V2]

X_train_v2 = X_v2.loc[X_train.index]
X_test_v2  = X_v2.loc[test_idx]

print(f'Train: {X_train_v2.shape} | Test: {X_test_v2.shape}')
```

```python
# Random Forest — Iteration 2
rf_v2 = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
rf_v2.fit(X_train_v2, y_train)
y_pred_rf_v2 = rf_v2.predict(X_test_v2)

bal_rf_v2 = balanced_accuracy_score(y_test, y_pred_rf_v2)
f1_rf_v2  = f1_score(y_test, y_pred_rf_v2, average='macro')

print('=== Random Forest — Iteration 2 ===')
print(f'Balanced Accuracy: {bal_rf_v2:.3f}  (Iter 1: {bal_rf_v1:.3f}, Δ={bal_rf_v2 - bal_rf_v1:+.3f})')
print(f'F1 (macro):        {f1_rf_v2:.3f}  (Iter 1: {f1_rf_v1:.3f}, Δ={f1_rf_v2 - f1_rf_v1:+.3f})')
print()
print(classification_report(y_test, y_pred_rf_v2, target_names=le_risk.classes_))
```

```python
# XGBoost — Iteration 2
sw_v2 = compute_sample_weight('balanced', y_train)

xgb_v2 = XGBClassifier(
    n_estimators=200,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    random_state=42,
    eval_metric='mlogloss',
    verbosity=0
)
xgb_v2.fit(X_train_v2, y_train, sample_weight=sw_v2)
y_pred_xgb_v2 = xgb_v2.predict(X_test_v2)

bal_xgb_v2 = balanced_accuracy_score(y_test, y_pred_xgb_v2)
f1_xgb_v2  = f1_score(y_test, y_pred_xgb_v2, average='macro')

print('=== XGBoost — Iteration 2 ===')
print(f'Balanced Accuracy: {bal_xgb_v2:.3f}  (Iter 1: {bal_xgb_v1:.3f}, Δ={bal_xgb_v2 - bal_xgb_v1:+.3f})')
print(f'F1 (macro):        {f1_xgb_v2:.3f}  (Iter 1: {f1_xgb_v1:.3f}, Δ={f1_xgb_v2 - f1_xgb_v1:+.3f})')
print()
print(classification_report(y_test, y_pred_xgb_v2, target_names=le_risk.classes_))
```

## 3. Iterationsvergleich & Visualisierung

```python
# Vergleichstabelle: Iteration 1 vs Iteration 2
results = pd.DataFrame({
    'Iteration': ['Iter 1', 'Iter 1', 'Iter 2', 'Iter 2'],
    'Modell':    ['Random Forest', 'XGBoost', 'Random Forest', 'XGBoost'],
    'Features':  [3, 3, 8, 8],
    'Balanced Accuracy': [bal_rf_v1, bal_xgb_v1, bal_rf_v2, bal_xgb_v2],
    'F1 Macro':          [f1_rf_v1,  f1_xgb_v1,  f1_rf_v2,  f1_xgb_v2]
}).round(3)

print('=== Übersicht: Iterationsvergleich ===')
print(results.to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

labels   = ['RF Iter1', 'XGB Iter1', 'RF Iter2', 'XGB Iter2']
bal_vals = [bal_rf_v1, bal_xgb_v1, bal_rf_v2, bal_xgb_v2]
f1_vals  = [f1_rf_v1,  f1_xgb_v1,  f1_rf_v2,  f1_xgb_v2]
colors   = ['#aec6cf', '#aec6cf', '#3498db', '#3498db']

axes[0].bar(labels, bal_vals, color=colors, edgecolor='white')
axes[0].set_title('Balanced Accuracy: Iter 1 vs Iter 2')
axes[0].set_ylim(0, 1)
axes[0].set_ylabel('Balanced Accuracy')
for i, v in enumerate(bal_vals):
    axes[0].text(i, v + 0.01, f'{v:.3f}', ha='center', fontsize=9)

axes[1].bar(labels, f1_vals, color=colors, edgecolor='white')
axes[1].set_title('F1 Macro: Iter 1 vs Iter 2')
axes[1].set_ylim(0, 1)
axes[1].set_ylabel('F1 Macro')
for i, v in enumerate(f1_vals):
    axes[1].text(i, v + 0.01, f'{v:.3f}', ha='center', fontsize=9)

plt.tight_layout()
plt.savefig(FIGURES_DIR / 'ml_01_iteration_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Confusion Matrices: bestes Modell pro Iteration
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

best_v1_pred = y_pred_rf_v1 if bal_rf_v1 >= bal_xgb_v1 else y_pred_xgb_v1
best_v2_pred = y_pred_rf_v2 if bal_rf_v2 >= bal_xgb_v2 else y_pred_xgb_v2
best_v1_name = 'Random Forest' if bal_rf_v1 >= bal_xgb_v1 else 'XGBoost'
best_v2_name = 'Random Forest' if bal_rf_v2 >= bal_xgb_v2 else 'XGBoost'

for ax, (name, iteration, y_pred) in zip(
    axes,
    [(best_v1_name, 1, best_v1_pred), (best_v2_name, 2, best_v2_pred)]
):
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=le_risk.classes_, yticklabels=le_risk.classes_)
    ax.set_title(f'{name} — Iteration {iteration}\nBalanced Acc: {balanced_accuracy_score(y_test, y_pred):.3f}')
    ax.set_xlabel('Predicted')
    ax.set_ylabel('True')

plt.tight_layout()
plt.savefig(FIGURES_DIR / 'ml_02_confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 5. Fehleranalyse

```python
errors = X_test_v2.copy()
errors['true']      = le_risk.inverse_transform(y_test)
errors['predicted'] = le_risk.inverse_transform(best_v2_pred)
errors['correct']   = errors['true'] == errors['predicted']

print(f'Gesamtfehlerrate: {(~errors["correct"]).mean():.1%}')
print()
print('Häufigste Fehler-Kombinationen:')
err_table = (
    errors[~errors['correct']]
    .groupby(['true', 'predicted'])
    .size()
    .reset_index(name='count')
    .sort_values('count', ascending=False)
)
print(err_table)
print()

# Kritische Fehler: high als low klassifiziert
critical = errors[(errors['true'] == 'high') & (errors['predicted'] == 'low')]
print(f'Kritische Fehler (high → low): {len(critical)}')
print('→ Diese Fälle würden im App durch den CV-Output (Bildanalyse) korrigiert.')
```

## 6. Bestes Modell speichern

```python
# Bestes Modell aus Iteration 2 wählen und speichern
if bal_rf_v2 >= bal_xgb_v2:
    best_model = rf_v2
    best_name  = 'Random Forest'
    best_bal   = bal_rf_v2
    best_f1    = f1_rf_v2
else:
    best_model = xgb_v2
    best_name  = 'XGBoost'
    best_bal   = bal_xgb_v2
    best_f1    = f1_xgb_v2

# Modell muss mit FEATURES_V2 arbeiten — App wird entsprechend angepasst
model_meta = {
    'model_name': best_name,
    'features':   FEATURES_V2,
    'balanced_accuracy': round(best_bal, 3),
    'f1_macro':   round(best_f1, 3)
}
import json
with open(MODELS_DIR / 'ml_model_meta.json', 'w') as f:
    json.dump(model_meta, f, indent=2)

joblib.dump(best_model, MODELS_DIR / 'ml_risk_classifier.pkl')

print(f'Gespeichert: ml_risk_classifier.pkl ({best_name})')
print(f'Balanced Accuracy: {best_bal:.3f}')
print(f'F1 Macro:          {best_f1:.3f}')
print(f'Features:          {FEATURES_V2}')
```

## Iteration 3: Zweite Datenquelle — ISIC 2020 integrieren

**Ziel:** Erweiterung des Trainingsdatensatzes mit ISIC 2020 Metadaten.
Das ISIC 2020 Dataset ist eine **andere, unabhängige Datenquelle** (Kaggle Challenge 2020) mit binären Labels.

| Quelle | Zeilen | Labels | Neue Fälle für 'high' |
|---|---|---|---|
| HAM10000 | 9'958 | 7 Klassen (mel, nv, bcc, ...) | 1'627 |
| ISIC 2020 | 33'126 | binär (benign/malignant) | 584 |
| **Kombiniert** | **~42'500** | low/medium/high | **~2'200** |

**Mapping ISIC 2020 → risk_level:** malignant → high | benign → low (kein medium, da ISIC 2020 binär)

**Mapping anatom_site → localization:** head/neck→face | torso→trunk | palms/soles→foot

**Modell:** XGBoost (Gewinner aus Iter 2) mit FEATURES_V2


```python
# ISIC 2020 laden
isic = pd.read_csv('../data/raw/isic2020_train.csv')
print(f'ISIC 2020: {isic.shape}')
print(isic[['sex', 'age_approx', 'anatom_site_general_challenge', 'benign_malignant']].head(3))

# Fehlende Werte entfernen
isic = isic.dropna(subset=['sex', 'age_approx', 'anatom_site_general_challenge'])
print(f'Nach Cleaning: {isic.shape}')

# risk_level mapping: malignant=high, benign=low
isic['risk_level'] = isic['benign_malignant'].map({'malignant': 'high', 'benign': 'low'})

# anatom_site -> localization (angleichen an HAM10000 Werte)
site_map = {
    'head/neck':       'face',
    'upper extremity': 'upper extremity',
    'lower extremity': 'lower extremity',
    'torso':           'trunk',
    'palms/soles':     'foot',
    'oral/genital':    'genital'
}
isic['localization'] = isic['anatom_site_general_challenge'].map(site_map).fillna('unknown')
isic['age'] = isic['age_approx']

print('Risikoverteilung ISIC 2020:')
print(isic['risk_level'].value_counts())
print('Lokalisierungsverteilung ISIC 2020:')
print(isic['localization'].value_counts())
```

```python
# ISIC 2020 Features mit bestehenden Encodern kodieren
# (Encoder wurden in Iteration 1 auf HAM10000 gefittet)
known_locs  = set(le_loc.classes_)
known_sexes = set(le_sex.classes_)

isic['loc_safe'] = isic['localization'].apply(lambda x: x if x in known_locs else le_loc.classes_[0])
isic['sex_safe'] = isic['sex'].apply(lambda x: x if x in known_sexes else le_sex.classes_[0])

isic['sex_enc'] = le_sex.transform(isic['sex_safe'])
isic['loc_enc'] = le_loc.transform(isic['loc_safe'])
isic['risk_enc'] = le_risk.transform(isic['risk_level'])

# Feature Engineering identisch zu Iteration 2
isic['age_group']             = np.where(isic['age'] < 30, 0, np.where(isic['age'] <= 60, 1, 2)).astype(int)
isic['is_face']               = isic['localization'].isin(['face']).astype(int)
isic['is_trunk']              = isic['localization'].isin(['back', 'abdomen', 'chest', 'trunk']).astype(int)
isic['is_extremity']          = isic['localization'].isin(['lower extremity', 'upper extremity', 'foot', 'hand']).astype(int)
isic['is_high_risk_location'] = isic['localization'].isin(['back', 'lower extremity']).astype(int)

# HAM10000 Trainingsdaten + ISIC 2020 zusammenfuehren
ham_train  = df.loc[X_train.index][FEATURES_V2 + ['risk_enc']].copy()
isic_train = isic[FEATURES_V2 + ['risk_enc']].copy()

combined = pd.concat([ham_train, isic_train], ignore_index=True)
print(f'HAM10000 Train:  {len(ham_train):>6} Zeilen')
print(f'ISIC 2020:       {len(isic_train):>6} Zeilen')
print(f'Kombiniert:      {len(combined):>6} Zeilen')
print()
print('Risikoverteilung kombinierter Datensatz:')
print(pd.Series(le_risk.inverse_transform(combined['risk_enc'])).value_counts())
```

```python
# XGBoost auf kombiniertem Datensatz
X_combined = combined[FEATURES_V2]
y_combined = combined['risk_enc']
sw_combined = compute_sample_weight('balanced', y_combined)

xgb_v3 = XGBClassifier(
    n_estimators=200,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    random_state=42,
    eval_metric='mlogloss',
    verbosity=0
)
xgb_v3.fit(X_combined, y_combined, sample_weight=sw_combined)

# Evaluation auf demselben HAM10000 Testset (fair: nur HAM10000 im Test)
y_pred_xgb_v3 = xgb_v3.predict(X_test_v2)

bal_xgb_v3 = balanced_accuracy_score(y_test, y_pred_xgb_v3)
f1_xgb_v3  = f1_score(y_test, y_pred_xgb_v3, average='macro')

print('=== XGBoost --- Iteration 3 (HAM10000 + ISIC 2020) ===')
print(f'Balanced Accuracy: {bal_xgb_v3:.3f}  (Iter 2: {bal_xgb_v2:.3f}, Delta={bal_xgb_v3 - bal_xgb_v2:+.3f})')
print(f'F1 (macro):        {f1_xgb_v3:.3f}  (Iter 2: {f1_xgb_v2:.3f}, Delta={f1_xgb_v3 - f1_xgb_v2:+.3f})')
print()
print(classification_report(y_test, y_pred_xgb_v3, target_names=le_risk.classes_))
```

```python
# Vollstaendiger Vergleich aller 3 Iterationen
results_all = pd.DataFrame({
    'Iteration':         ['Iter 1', 'Iter 1', 'Iter 2', 'Iter 2', 'Iter 3'],
    'Modell':            ['Random Forest', 'XGBoost', 'Random Forest', 'XGBoost', 'XGBoost'],
    'Trainingsdaten':    ['HAM10000', 'HAM10000', 'HAM10000', 'HAM10000', 'HAM10000 + ISIC 2020'],
    'Features':          [3, 3, 8, 8, 8],
    'Balanced Accuracy': [bal_rf_v1, bal_xgb_v1, bal_rf_v2, bal_xgb_v2, bal_xgb_v3],
    'F1 Macro':          [f1_rf_v1,  f1_xgb_v1,  f1_rf_v2,  f1_xgb_v2,  f1_xgb_v3]
}).round(3)

print('=== Vollstaendiger Iterationsvergleich ===')
print(results_all.to_string(index=False))

# Balkendiagramm
fig, ax = plt.subplots(figsize=(13, 4))
labels_all = ['RF Iter1', 'XGB Iter1', 'RF Iter2', 'XGB Iter2', 'XGB Iter3\n(+ISIC2020)']
colors_all = ['#aec6cf', '#aec6cf', '#3498db', '#3498db', '#e74c3c']
x_pos = list(range(len(labels_all)))
w = 0.35
ax.bar([i - w/2 for i in x_pos], results_all['Balanced Accuracy'], w,
       label='Balanced Accuracy', color=colors_all, alpha=0.9)
ax.bar([i + w/2 for i in x_pos], results_all['F1 Macro'], w,
       label='F1 Macro', color=colors_all, alpha=0.5)
ax.set_xticks(x_pos)
ax.set_xticklabels(labels_all)
ax.set_ylim(0, 1)
ax.set_title('Vergleich alle Iterationen: ML Risikoklassifikation')
ax.legend()
plt.tight_layout()
plt.savefig('../reports/figures/ml_04_all_iterations.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Interpretation Iteration 3

- **ISIC 2020 liefert ~4x mehr Daten**, aber hauptsächlich 'low'- und 'high'-Fälle
- **Keine 'medium'-Klasse** in ISIC 2020 (binäres Dataset) — Klassifikation von akiec wird nicht verbessert
- Leichte Performance-Veränderung zeigt: mehr Metadaten alleine genügen nicht — das Bild (CV-Block) ist entscheidend
- **Integration klar dokumentiert:** ML-Block verwendet 2 Datenquellen mit unterschiedlicher Herkunft und Struktur

**Fazit:** Das beste finale Modell für die App bleibt XGBoost Iter 2 (trainiert auf HAM10000).
ISIC 2020 demonstriert die Multi-Source-Integration und liefert zusätzliche 'high'-Fälle als Trainingsboost.
